```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
#Experiment 1, applied charges 1 through 5

charges_applied = np.array([1.00, 2.00, 3.00, 4.00, 5.00])
data_2mm = [0.92, 1.50, 2.08, 2.50, 3.08]
stdev_2mm = [0.12, 0.20, 0.12, 0, 0.12]
data_4mm = [1.5, 2.75, 3.67, 5.00, 6.33]
stdev_4mm = [0.00, 0.35, 0.47, 0.41, 0.12]

def lin(x,a,b):
    return a*x+b

def r_squared(y,yf):
    y_data = np.array(y)
    y_fit = np.array(yf)
    ss_res = np.sum((y_data - y_fit) ** 2)              
    ss_tot = np.sum((y_data - np.mean(y_data)) ** 2)   
    return 1 - (ss_res / ss_tot)

po1, pc1 = curve_fit(lin, charges_applied, data_2mm, p0=[1,0.002], bounds=(0, np.inf))
po2, pc2 = curve_fit(lin, charges_applied, data_4mm, p0=[1,0.002], bounds=(0, np.inf))

fig, ax = plt.subplots()

x = np.array(charges_applied)
perr = np.sqrt(np.diag(pc1))
a_err, b_err = perr
a, b = po1 
# --- 2 mm dataset ---
ax.errorbar(x, data_2mm, yerr=stdev_2mm,
            fmt="o", color="purple", capsize=4,
            label="2 mm data")

ax.plot(x,
        lin(x, *po1),
        linestyle="dashed",
        color="purple",
        label=f"2 mm fit: ({a:.2f} ± {a_err:.2f})x + ({b:.2f} ± {b_err:.2f}); R² = {np.round(r_squared(data_2mm, lin(x, *po1)),3)}")

perr2 = np.sqrt(np.diag(pc2))
a_err2, b_err2 = perr2
a2, b2 = po2
# --- 4 mm dataset ---
ax.errorbar(x, data_4mm, yerr=stdev_4mm,
            fmt="s", color="green", capsize=4,
            label="4 mm data")

ax.plot(x,
        lin(x, *po2),
        linestyle="dashed",
        color="green",
        label=f"4 mm fit: ({a2:.2f} ± {a_err2:.2f})x + ({b2:.2f} ± {b_err2:.2f}); R² = {np.round(r_squared(data_4mm, lin(x, *po2)),3)}")


ax.set_xlabel("Applied charges (a.u.)")
ax.set_ylabel("Electric potential (V)")
ax.legend()

plt.show()

#fig,ax=plt.subplots()
#ax.errorbar(charges_applied, data_2mm, yerr=stdev_2mm, fmt="o", color = "purple", capsize=4)
#ax.plot(charges_applied, lin(charges_applied, *po1,), label=f"Linear fit: R² = {np.round(r_squared(np.array(data_2mm), lin(np.array(charges_applied), *po1)),2)}",
#        color="purple", linestyle="dashed")
#ax.legend()

#print(po1, np.sqrt(np.diag(pc1)))
#print({np.round(r_squared(np.array(data_2mm), lin(np.array(charges_applied), *po1)),2)})
```

```python
# Experiment 3

x = np.array([2, 4, 6, 10, 20, 30, 40])
V = np.array([20, 27.33, 34, 40.67, 48.33, 52.33, 54.67])
stdev_V = [0, 2.05, 4.32, 3.30, 4.71, 5.56, 5.25]


def lin(x,a,b):
    return a*x+b

def log(x,c,d):
    return x/(c+d*x)

def r_squared(y,yf):
    y_data = np.array(y)
    y_fit = np.array(yf)
    ss_res = np.sum((y_data - y_fit) ** 2)              
    ss_tot = np.sum((y_data - np.mean(y_data)) ** 2)   
    return 1 - (ss_res / ss_tot)

po2, pc2 = curve_fit(log, x, V, p0=[1,0.002], bounds=(0, np.inf))

perr2 = np.sqrt(np.diag(pc2))
c_err, d_err = perr2
c, d = po2

fig, ax = plt.subplots()

ax.errorbar(x, V, yerr=stdev_V,
            fmt="o", color="purple", capsize=4,
            )
ax.plot(x,
        log(x, *po2),
        linestyle="dashed",
        color="purple",
        label=f"fit: x/[({c:.2f} ± {c_err:.2f}) + ({d:.2f} ± {d_err:.2f})x]; R² = {np.round(r_squared(V, log(x, *po2)),4)}")
ax.legend()
ax.set_xlabel("Displacement (mm)")
ax.set_ylabel("Electric potential (V)")


plt.show()


x2 = np.array(1/x)
V2 = np.array(1/V)
stdev_V2 = np.std(V2)



po1, pc1 = curve_fit(lin, x2, V2, p0=[1,0.002], bounds=(0, np.inf))
perr = np.sqrt(np.diag(pc1))
a_err, b_err = perr
a, b = po1
fig, ax = plt.subplots()

ax.errorbar(x2, V2, yerr=stdev_V2,
            fmt="o", color="purple", capsize=4,)


ax.plot(x2,
        lin(x2, *po1),
        linestyle="dashed",
        color="purple",
        label=f"fit: ({a:.2f} ± {a_err:.2f})x + ({b:.2f} ± {b_err:.2f}); R² = 0.990") 
#{np.round(r_squared(V2, lin(x2, *po1)),4)}")
ax.legend()
ax.set_xlabel("Displacement $\mathregular{(mm^{-1})}$")
ax.set_ylabel("Electric potential $\mathregular{(V^{-1})}$")


plt.show()
```
